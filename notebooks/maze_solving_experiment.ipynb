```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../results/maze_results.csv")

comparison = df.groupby("algorithm")[
    ["path_length", "nodes_expanded"]
].mean()

comparison.plot(kind="bar", figsize=(10, 5))

plt.title("BFS vs DFS vs A* — Performance Comparison")
plt.xlabel("Algorithm")
plt.ylabel("Average Value")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
comparison["path_length"].plot(kind="bar", title="Average Path Length by Algorithm")
plt.ylabel("Average Path Length")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

comparison["nodes_expanded"].plot(kind="bar", title="Average Nodes Expanded by Algorithm")
plt.ylabel("Average Nodes Expanded")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
